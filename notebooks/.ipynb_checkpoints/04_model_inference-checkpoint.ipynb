```python
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder

class SolarFaultLSTM(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, num_classes):
        super(SolarFaultLSTM, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        out, _ = self.lstm(x)
        out = self.fc(out[:, -1, :])
        return out

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = SolarFaultLSTM(input_size=4, hidden_size=64, num_layers=2, num_classes=4).to(device)

weights_path = '/content/solar_lstm_weights.pth'
model.load_state_dict(torch.load(weights_path, map_location=device))
model.eval()

df = pd.read_csv('/content/labeled_fault_data.xls', parse_dates=['timestamp'])
df = df.dropna().sort_values('timestamp').reset_index(drop=True)

label_encoder = LabelEncoder()
df['label_encoded'] = label_encoder.fit_transform(df['label'])

features = ['v_dc', 'i_dc', 'irradiance', 'temp_ambient']
scaler = StandardScaler()
df[features] = scaler.fit_transform(df[features])

# 3. Simulating Live Telemetry
seq_length = 60
sample_data = df[features].values[-seq_length:]
actual_label = df['label'].values[-1]

sample_tensor = torch.tensor(sample_data, dtype=torch.float32).unsqueeze(0).to(device)

with torch.no_grad():
    prediction = model(sample_tensor)
    predicted_idx = torch.argmax(prediction, dim=1).item()
    predicted_label = label_encoder.inverse_transform([predicted_idx])[0]

print(f"LIVE INFERENCE TEST")
print(f"Actual Physical State:   {actual_label}")
print(f"Model Predicted State:   {predicted_label}")
```
